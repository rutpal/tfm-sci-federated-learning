```python
import requests
import json

url = "https://api.boavizta.org/v1/terminal/laptop"

payload = {
    "model": {
        "type": "laptop"
    }
}

response = requests.post(
    url,
    json=payload
)

print(response.status_code)

data = response.json()

print(json.dumps(data, indent=2))
```
