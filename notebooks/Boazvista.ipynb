```python
import requests
import pandas as pd
from datetime import date

BASE_URL = "https://api.boavizta.org/v1"


retrieval_date = date.today().isoformat()

def get_terminal_embodied(device_type):

    response = requests.get(
        f"{BASE_URL}/terminal/{device_type}?verbose=true"
    )

    if not response.ok:
        return None

    data = response.json()

    return {

    "device_category": device_type,

    "te_kgco2eq":
        data["impacts"]["gwp"]["embedded"]["value"],

    "te_min_kgco2eq":
        data["impacts"]["gwp"]["embedded"]["min"],

    "te_max_kgco2eq":
        data["impacts"]["gwp"]["embedded"]["max"],

    "el_hours":
        data["verbose"]["duration"]["value"],

    "el_years":
        data["verbose"]["duration"]["value"]
        / 24
        / 365,

    "source": "boavizta",
    
    "retrieval_date": retrieval_date
}
```

```python
devices = [
    "laptop",
    "desktop",
    "smartphone",
    "tablet"
]
```

```python
records = []

for device in devices:

    result = get_terminal_embodied(
        device
    )

    if result:
        records.append(result)
```

```python
df = pd.DataFrame(records)

df
```

Server

```python
response = requests.get(
    f"{BASE_URL}/server/"
)

if response.ok:

    data = response.json()

    records.append({

        "device_category": "server",

        # TE (SCI)
        "te_kgco2eq":
            data["impacts"]["gwp"]["embedded"]["value"],

        "te_min_kgco2eq":
            data["impacts"]["gwp"]["embedded"]["min"],

        "te_max_kgco2eq":
            data["impacts"]["gwp"]["embedded"]["max"],

        # EL (SCI)
        "el_hours":
            data["verbose"]["hours_life_time"]["value"],

        "el_years":
            data["verbose"]["hours_life_time"]["value"]
            / 24
            / 365,

        "source":
            "boavizta",
            
        "retrieval_date": retrieval_date
    })
```

```python
df.to_csv(
    "boavizta_reference.csv",
    index=False
)

df
```
